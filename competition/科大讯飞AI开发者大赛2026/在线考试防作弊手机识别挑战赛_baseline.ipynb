```python
model = YOLO("runs/detect/train/weights/best.pt")
```

```python
import os
import glob
from PIL import Image
import json

os.makedirs('submit', exist_ok=True)
test_paths = glob.glob('test/images/*')

# 准备提交数据
submission_data = []

for path in test_paths:
    try:
        # 获取图片尺寸
        img = Image.open(path)
        img_width, img_height = img.size
        
        # 运行模型预测
        predictions = model(path, save_txt=None)
        
        # 准备检测结果
        detections = []
        for idx, prediction in enumerate(predictions[0].boxes.xywhn):
            cls = int(predictions[0].boxes.cls[idx].item())
            confidence = predictions[0].boxes.conf[idx].item()
            
            # 只保留phone类别（假设phone的类别索引为0，请根据实际情况调整）
            if cls == 0:  # 如果phone不是类别0，请修改为正确的类别索引
                # 将归一化坐标转换为像素坐标 [x_min, y_min, x_max, y_max]
                x_center = prediction[0].item() * img_width
                y_center = prediction[1].item() * img_height
                width = prediction[2].item() * img_width
                height = prediction[3].item() * img_height
                
                x_min = int(x_center - width / 2)
                y_min = int(y_center - height / 2)
                x_max = int(x_center + width / 2)
                y_max = int(y_center + height / 2)
                
                # 确保坐标在图片范围内
                x_min = max(0, min(x_min, img_width - 1))
                y_min = max(0, min(y_min, img_height - 1))
                x_max = max(x_min + 1, min(x_max, img_width))
                y_max = max(y_min + 1, min(y_max, img_height))
                
                detections.append({
                    "bbox": [x_min, y_min, x_max, y_max],
                    "confidence": round(confidence, 6),
                    "class": "phone"
                })
        
        # 添加到提交数据
        submission_data.append({
            "image_id": os.path.basename(path),
            "detections": detections
        })
        
    except Exception as e:
        print(f"Error processing {path}: {e}")
        # 出错时添加空检测
        submission_data.append({
            "image_id": os.path.basename(path),
            "detections": []
        })

# 保存为JSON格式
with open("./submit/predictions.json", "w", encoding="utf-8") as f:
    json.dump(submission_data, f, ensure_ascii=False, indent=2)

print(f"Submission saved to ./submit/predictions.json")
print(f"Total images processed: {len(submission_data)}")
```
